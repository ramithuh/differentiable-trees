```python
import wandb
api = wandb.Api()
```

```python
runs = api.runs("ramithx1/Batch-Run-Small-Parsimony-A100")
```

```python
# for run in runs:
#     print(run.name)
```

```python
errors = {}

sankoffs = {}
elves = {}

for l in [1024,512,256,128,64,32,16]:
    errors[str(l)] = []
    sankoffs[str(l)] = []
    elves[str(l)] = []
    
    for run in runs:
        if(run.config['n_leaves'] == l and run.config['tLs']==[0, 0.01, 100, 3]):
            
            # print(run.config['epochs'])
            
            cost = run.summaryMetrics['traversal cost']
            gt   = run.summaryMetrics['sankoff_cost'] 


            errors[str(l)].append(cost - gt)
            sankoffs[str(l)].append(gt)
            elves[str(l)].append(cost)
            

            # print(run.summaryMetrics.keys())
            # print(run.summaryMetrics['sankoff_cost'])
            # print(run.summaryMetrics['traversal cost'])
            # print(run.history)
        
```

```python
for key in [1024,512,256,128,64,32,16][::-1]:
    err = np.array(errors[str(key)])
    s_gt = np.array(sankoffs[str(key)])
    sol  = np.array(elves[str(key)])
    
    print(key, end = " -> ")
    assert err.shape == sol.shape
    assert s_gt.shape == sol.shape
    print(f" average sankoff = {s_gt.mean()}, avg ans = {sol.mean()}, avg error = {err.mean()}, % error = {err.mean()/s_gt.mean()*100}") 

    
    
```
